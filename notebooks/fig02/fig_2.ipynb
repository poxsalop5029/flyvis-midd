```python
if False:  # acquisition archive; no new inference
    ## #cell 003
    # Re-infer the 50 official networks × 15 luminance levels × Original/Mirror.
    # Uses the same 50-frame static input, BoxEye, fade-in, stable window, and CW-positive metric as Appendix 01.
    
    import gc
    from pathlib import Path
    import numpy as np
    import pandas as pd
    import torch
    import flyvis
    from flyvis.datasets.rendering import BoxEye
    from flyvis_midd import build_3midd_sequences, render_sector_image
    from flyvis_midd.flow import run_flow_inference
    from flyvis_midd.evaluation import CANONICAL_MIDD, analyze_midd_flow, aggregate_midd_metrics
    CONFIG = CANONICAL_MIDD
    
    REPO_ROOT = Path(REPO_ROOT) if "REPO_ROOT" in globals() else next(
        p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents, Path("/content/flyvis-midd"), Path("/content"))
        if (p / "pyproject.toml").is_file() and (p / "src" / "flyvis_midd").is_dir()
    )
    DEVICE = torch.device("cpu")
    LEVELS = [16, 32, 48, 64, 80, 96, 112, 128, 143, 159, 175, 191, 207, 223, 239]
    VARIANTS = ("original", "mirror")
    MODEL_IDS = [f"flow/0000/{i:03d}" for i in range(50)]
    N_FRAMES, DT, FADE = CONFIG.n_frames, CONFIG.dt, CONFIG.fade_seconds
    
    output_dir = REPO_ROOT / "data/precomputed/fig02"
    output_dir.mkdir(parents=True, exist_ok=True)
    raw_path = output_dir / "fig2B_luminance_tangent_per_model_condition.csv"
    summary_path = output_dir / "fig2B_luminance_A_per_model.csv"
    model_paths = {name: flyvis.results_dir / name for name in MODEL_IDS}
    missing = [str(path) for path in model_paths.values() if not path.is_dir()]
    if missing:
        raise FileNotFoundError(
            f"Missing {len(missing)} pretrained models. Run the FlyVis setup/download cell first; "
            f"examples: {missing[:3]}"
        )
    
    eye = BoxEye(extent=CONFIG.eye_extent, kernel_size=CONFIG.eye_kernel_size)
    centers_yx = eye.receptor_centers.detach().cpu().numpy()
    inputs = {}
    for c in LEVELS:
        sequences = build_3midd_sequences(c)
        for variant in VARIANTS:
            image, _ = render_sector_image(
                sequences[variant], image_size=382, inner_diameter=55,
                outer_diameter=382, phase_deg=0.0,
            )
            movie = np.repeat((image.astype(np.float32) / 255)[None, None], N_FRAMES, axis=1)
            inputs[(c, variant)] = eye(torch.as_tensor(movie, dtype=torch.float32)).cpu()
    assert all(tuple(x.shape) == (1, N_FRAMES, 1, 721) for x in inputs.values())
    
    columns = ["model_id", "C", "variant", "T_cw_positive", "mean_flow_magnitude"]
    lum_raw = pd.read_csv(raw_path) if raw_path.is_file() else pd.DataFrame(columns=columns)
    if len(lum_raw):
        lum_raw["C"] = lum_raw["C"].astype(int)
    done = set(map(tuple, lum_raw[["model_id", "C", "variant"]].itertuples(index=False, name=None)))
    
    for model_number, model_id in enumerate(MODEL_IDS, 1):
        pending = [(c, v) for c in LEVELS for v in VARIANTS if (model_id, c, v) not in done]
        if not pending:
            print(f"[{model_number:02d}/50] {model_id}: complete")
            continue
        view = flyvis.NetworkView(model_paths[model_id])
        network = view.init_network().to(DEVICE).eval()
        decoder = view.init_decoder()["flow"].to(DEVICE).eval()
        for c, variant in pending:
            result = run_flow_inference(
                inputs[(c, variant)], network, decoder, dt=DT, fade=FADE, device=DEVICE,
            )
            flow = result["flow"][0].numpy()
            assert flow.shape == (N_FRAMES, 2, 721) and np.isfinite(flow).all()
            metric = analyze_midd_flow(result["flow"], centers_yx, config=CONFIG)
            lum_raw.loc[len(lum_raw)] = [model_id, c, variant,
                                         metric["mean_signed_tangent"], metric["mean_flow_magnitude"]]
            done.add((model_id, c, variant))
            print(f"[{model_number:02d}/50] C={c:03d} {variant}")
        temp_path = raw_path.with_suffix(".tmp")
        lum_raw.sort_values(["model_id", "C", "variant"]).to_csv(temp_path, index=False)
        temp_path.replace(raw_path)
        del decoder, network, view
        gc.collect()
    
    assert len(lum_raw) == 1500 and not lum_raw.duplicated(["model_id", "C", "variant"]).any()
    assert lum_raw["model_id"].nunique() == 50
    lum_a = lum_raw.pivot(index=["model_id", "C"], columns="variant", values="T_cw_positive").reset_index()
    pair_metrics = aggregate_midd_metrics({"original": lum_a["original"], "mirror": lum_a["mirror"]})
    for key in ("A", "abs_A", "actual_reversal"):
        lum_a[key] = pair_metrics[key]
    lum_a.sort_values(["model_id", "C"]).to_csv(summary_path, index=False)
    print(f"Inference complete: {len(lum_raw)}/1500; saved {raw_path.name} and {summary_path.name}")
```

```python
if False:  # acquisition archive; no new inference
    ## #cell 006
    # Fig. 2D — 4MIDD response across the official-50 ensemble
    # Same cohort / BoxEye / timing / CW-positive metric as Fig. 2B–C.
    # Fixed condition: 4MIDD C1=85, C2=170.
    #
    # Core operations are reused from flyvis_midd:
    #   build_4midd_sequences
    #   render_sector_image
    #   run_flow_inference
    #   analyze_midd_flow / aggregate_midd_metrics
    
    import gc
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    import torch
    import flyvis
    
    from flyvis_midd import build_4midd_sequences, render_sector_image
    from flyvis_midd.flow import run_flow_inference
    from flyvis_midd.evaluation import CANONICAL_MIDD, analyze_midd_flow, aggregate_midd_metrics
    CONFIG = CANONICAL_MIDD
    
    
    # ---------------------------------------------------------------------
    # 1. Fixed 4MIDD condition used in Appendix 01 / official-50 analysis
    # ---------------------------------------------------------------------
    
    C1, C2 = 85, 170
    
    from pathlib import Path
    from flyvis.datasets.rendering import BoxEye
    
    # Keep this cell runnable after a kernel restart, using the same settings as Fig. 2B.
    DEVICE = torch.device("cpu")
    MODEL_IDS = [f"flow/0000/{i:03d}" for i in range(50)]
    N_FRAMES, DT, FADE = CONFIG.n_frames, CONFIG.dt, CONFIG.fade_seconds
    REPO_ROOT = Path(REPO_ROOT) if "REPO_ROOT" in globals() else next(
        p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents, Path("/content/flyvis-midd"), Path("/content"))
        if (p / "pyproject.toml").is_file() and (p / "src" / "flyvis_midd").is_dir()
    )
    output_dir = REPO_ROOT / "data/precomputed/fig02"
    output_dir.mkdir(parents=True, exist_ok=True)
    model_paths = {name: flyvis.results_dir / name for name in MODEL_IDS}
    missing = [str(path) for path in model_paths.values() if not path.is_dir()]
    if missing:
        raise FileNotFoundError(
            f"Missing {len(missing)} pretrained models. Run the FlyVis setup/download cell first; "
            f"examples: {missing[:3]}"
        )
    eye = BoxEye(extent=CONFIG.eye_extent, kernel_size=CONFIG.eye_kernel_size)
    centers_yx = eye.receptor_centers.detach().cpu().numpy()
    VARIANTS_4MIDD = ("original", "mirror")
    
    # Resolve the paths explicitly so this cell also works when run on its own.
    REPO_ROOT = Path(REPO_ROOT) if "REPO_ROOT" in globals() else next(
        p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents, Path("/content/flyvis-midd"), Path("/content"))
        if (p / "pyproject.toml").is_file() and (p / "src" / "flyvis_midd").is_dir()
    )
    output_dir = REPO_ROOT / "data/precomputed/fig02"
    output_dir.mkdir(parents=True, exist_ok=True)
    raw_path_4 = output_dir / "fig2D_4MIDD_tangent_per_model_condition.csv"
    summary_path_4 = output_dir / "fig2D_4MIDD_A_per_model.csv"
    
    
    # ---------------------------------------------------------------------
    # 2. Build Original / Mirror with the existing flyvis_midd API
    # ---------------------------------------------------------------------
    
    sequences_4 = build_4midd_sequences(C1, C2)
    
    inputs_4 = {}
    
    for variant in VARIANTS_4MIDD:
        image, _ = render_sector_image(
            sequences_4[variant],
            image_size=382,
            inner_diameter=55,
            outer_diameter=382,
            phase_deg=0.0,
        )
    
        movie = np.repeat(
            (image.astype(np.float32) / 255.0)[None, None],
            N_FRAMES,
            axis=1,
        )
    
        inputs_4[variant] = eye(
            torch.as_tensor(movie, dtype=torch.float32)
        ).cpu()
    
    assert all(
        tuple(x.shape) == (1, N_FRAMES, 1, 721)
        for x in inputs_4.values()
    )
    
    
    # ---------------------------------------------------------------------
    # 3. Official-50 inference
    # ---------------------------------------------------------------------
    
    columns = [
        "model_id",
        "variant",
        "T_cw_positive",
        "mean_flow_magnitude",
    ]
    
    four_raw = (
        pd.read_csv(raw_path_4)
        if raw_path_4.is_file()
        else pd.DataFrame(columns=columns)
    )
    
    done = set(
        map(
            tuple,
            four_raw[["model_id", "variant"]]
            .itertuples(index=False, name=None),
        )
    )
    
    
    for model_number, model_id in enumerate(MODEL_IDS, 1):
    
        pending = [
            variant
            for variant in VARIANTS_4MIDD
            if (model_id, variant) not in done
        ]
    
        if not pending:
            print(f"[{model_number:02d}/50] {model_id}: complete")
            continue
    
        view = flyvis.NetworkView(model_paths[model_id])
        network = view.init_network().to(DEVICE).eval()
        decoder = view.init_decoder()["flow"].to(DEVICE).eval()
    
        for variant in pending:
    
            result = run_flow_inference(
                inputs_4[variant],
                network,
                decoder,
                dt=DT,
                fade=FADE,
                device=DEVICE,
            )
    
            flow = result["flow"][0].numpy()
    
            assert flow.shape == (N_FRAMES, 2, 721)
            assert np.isfinite(flow).all()
    
            metric = analyze_midd_flow(result["flow"], centers_yx, config=CONFIG)
    
            four_raw.loc[len(four_raw)] = [
                model_id,
                variant,
                metric["mean_signed_tangent"],
                metric["mean_flow_magnitude"],
            ]
    
            done.add((model_id, variant))
    
            print(
                f"[{model_number:02d}/50] "
                f"4MIDD {C1}/{C2} {variant}"
            )
    
        temp_path = raw_path_4.with_suffix(".tmp")
    
        four_raw.sort_values(
            ["model_id", "variant"]
        ).to_csv(temp_path, index=False)
    
        temp_path.replace(raw_path_4)
    
        del decoder, network, view
        gc.collect()
    
    
    # ---------------------------------------------------------------------
    # 4. Model-wise Original / Mirror summary
    # ---------------------------------------------------------------------
    
    assert len(four_raw) == 50 * 2
    assert four_raw["model_id"].nunique() == 50
    assert not four_raw.duplicated(["model_id", "variant"]).any()
    
    four_a = (
        four_raw
        .pivot(
            index="model_id",
            columns="variant",
            values="T_cw_positive",
        )
        .reset_index()
    )
    
    pair_metrics = aggregate_midd_metrics({"original": four_a["original"], "mirror": four_a["mirror"]})
    for key in ("A", "abs_A", "actual_reversal"):
        four_a[key] = pair_metrics[key]
    
    four_a.sort_values("model_id").to_csv(
        summary_path_4,
        index=False,
    )
    
    
    # ---------------------------------------------------------------------
    # 5. Fig. 2D
    # ---------------------------------------------------------------------
    
    reversal = four_a["actual_reversal"]
    
    limit = (
        1.05
        * np.abs(
            four_a[["original", "mirror"]].to_numpy()
        ).max()
    )
    
    fig, ax = plt.subplots(figsize=(3.8, 3.6))
    
    ax.scatter(
        four_a.loc[~reversal, "original"],
        four_a.loc[~reversal, "mirror"],
        color="0.65",
        s=28,
        alpha=0.8,
        label="Same sign",
    )
    
    ax.scatter(
        four_a.loc[reversal, "original"],
        four_a.loc[reversal, "mirror"],
        color="tab:blue",
        s=30,
        alpha=0.9,
        label="Actual reversal",
    )
    
    ax.axhline(0, color="0.5", linewidth=0.8)
    ax.axvline(0, color="0.5", linewidth=0.8)
    
    # A = 0 line: Original = Mirror
    ax.plot(
        [-limit, limit],
        [-limit, limit],
        "--",
        color="0.8",
        linewidth=0.8,
    )
    
    ax.set(
        xlim=(-limit, limit),
        ylim=(-limit, limit),
        aspect="equal",
        title="4MIDD  C1 = 85, C2 = 170",
    )
    
    ax.set_xlabel(
        r"$T_{\mathrm{Original}}$ (CW positive)"
    )
    ax.set_ylabel(
        r"$T_{\mathrm{Mirror}}$ (CW positive)"
    )
    
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    
    ax.legend(
        frameon=False,
        fontsize=8,
        loc="best",
    )
    
    fig.tight_layout()
    plt.show()
    
    
    # ---------------------------------------------------------------------
    # 6. Numerical QA against Appendix 01
    # ---------------------------------------------------------------------
    
    print("Models:", four_a["model_id"].nunique())
    print(f"Mean A: {four_a['A'].mean():+.8f}")
    print(f"SEM A : {four_a['A'].sem():.8f}")
    print(
        "Actual reversal:",
        f"{int(four_a['actual_reversal'].sum())}/50",
    )
```

```python
if False:  # acquisition archive; no new inference
    ## #cell 008
    # Fig. 2F — Direction agreement across the official-50 ensemble
    #
    # Reference direction:
    #   sign of the 50-model mean A for each stimulus.
    #
    # This is an ensemble-internal direction measure.
    # It is NOT a comparison with human perceptual direction.
    #
    # Shows:
    #   1. all 50 models
    #   2. models showing Actual reversal only
    
    from pathlib import Path
    
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    
    
    # ---------------------------------------------------------------------
    # 1. Load the matched official-50 results
    # ---------------------------------------------------------------------
    
    repo_root = Path(REPO_ROOT)
    data_dir = repo_root / "data/precomputed/fig02"
    
    lum_a = pd.read_csv(
        data_dir / "fig2B_luminance_A_per_model.csv"
    )
    
    four_a = pd.read_csv(
        data_dir / "fig2D_4MIDD_A_per_model.csv"
    )
    
    assert lum_a["model_id"].nunique() == 50
    assert four_a["model_id"].nunique() == 50
    
    
    # ---------------------------------------------------------------------
    # 2. Assemble the three fixed MIDD conditions
    # ---------------------------------------------------------------------
    
    groups = {
        "3MIDD 48": (
            lum_a[lum_a["C"] == 48]
            .copy()
        ),
        "3MIDD 207": (
            lum_a[lum_a["C"] == 207]
            .copy()
        ),
        "4MIDD 85/170": (
            four_a.copy()
        ),
    }
    
    rows = []
    
    for condition, q in groups.items():
    
        assert len(q) == 50
        assert q["model_id"].nunique() == 50
    
        # Ensemble reference direction for this stimulus
        mean_A = q["A"].mean()
        reference_sign = np.sign(mean_A)
    
        assert reference_sign != 0
    
        same_direction = (
            np.sign(q["A"])
            == reference_sign
        )
    
        # Actual reversal:
        # Original and Mirror have opposite signed rotation
        actual_reversal = aggregate_midd_metrics({"original": q["original"], "mirror": q["mirror"]})["actual_reversal"]
    
        reversal_q = q.loc[actual_reversal]
    
        reversal_same_direction = (
            np.sign(reversal_q["A"])
            == reference_sign
        )
    
        rows.append(
            {
                "condition": condition,
                "mean_A": mean_A,
                "reference_sign": int(reference_sign),
    
                "n_all": len(q),
                "n_same_all": int(
                    same_direction.sum()
                ),
                "agreement_all": float(
                    same_direction.mean()
                ),
    
                "n_reversal": len(reversal_q),
                "n_same_reversal": int(
                    reversal_same_direction.sum()
                ),
                "agreement_reversal": float(
                    reversal_same_direction.mean()
                ),
            }
        )
    
    
    agreement = pd.DataFrame(rows)
    
    
    # ---------------------------------------------------------------------
    # 3. Reproduction QA against Appendix 01
    # ---------------------------------------------------------------------
    
    expected = {
        "3MIDD 48": {
            "all": (43, 50),
            "reversal": (20, 21),
        },
        "3MIDD 207": {
            "all": (40, 50),
            "reversal": (17, 18),
        },
        "4MIDD 85/170": {
            "all": (34, 50),
            "reversal": (7, 12),
        },
    }
    
    for _, row in agreement.iterrows():
    
        exp = expected[row["condition"]]
    
        assert (
            int(row["n_same_all"]),
            int(row["n_all"]),
        ) == exp["all"]
    
        assert (
            int(row["n_same_reversal"]),
            int(row["n_reversal"]),
        ) == exp["reversal"]
    
    
    # ---------------------------------------------------------------------
    # 4. Fig. 2F
    # ---------------------------------------------------------------------
    
    condition_order = [
        "3MIDD 48",
        "3MIDD 207",
        "4MIDD 85/170",
    ]
    
    plot_df = (
        agreement
        .set_index("condition")
        .loc[condition_order]
        .reset_index()
    )
    
    x = np.arange(
        len(condition_order)
    )
    
    width = 0.34
    
    fig, ax = plt.subplots(
        figsize=(5.4, 3.7)
    )
    
    bars_all = ax.bar(
        x - width / 2,
        100 * plot_df["agreement_all"],
        width,
        color="0.65",
        label="All models",
    )
    
    bars_reversal = ax.bar(
        x + width / 2,
        100 * plot_df["agreement_reversal"],
        width,
        color="tab:blue",
        label="Actual-reversal models",
    )
    
    # 50% descriptive reference
    ax.axhline(
        50,
        color="0.55",
        linestyle="--",
        linewidth=0.9,
    )
    
    ax.set_xticks(x)
    ax.set_xticklabels(
        condition_order
    )
    
    ax.set_ylim(
        0,
        105,
    )
    
    ax.set_ylabel(
        "Direction agreement (%)"
    )
    
    ax.set_xlabel(
        "Stimulus"
    )
    
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    
    ax.legend(
        frameon=False,
        fontsize=8,
    )
    
    
    # Counts above bars
    for bar, (_, row) in zip(
        bars_all,
        plot_df.iterrows(),
    ):
        ax.text(
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height() + 2,
            f"{int(row['n_same_all'])}/{int(row['n_all'])}",
            ha="center",
            va="bottom",
            fontsize=8,
        )
    
    for bar, (_, row) in zip(
        bars_reversal,
        plot_df.iterrows(),
    ):
        ax.text(
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height() + 2,
            (
                f"{int(row['n_same_reversal'])}/"
                f"{int(row['n_reversal'])}"
            ),
            ha="center",
            va="bottom",
            fontsize=8,
        )
    
    fig.tight_layout()
    plt.show()
    
    
    # ---------------------------------------------------------------------
    # 5. Numerical summary
    # ---------------------------------------------------------------------
    
    display_columns = [
        "condition",
        "mean_A",
        "n_same_all",
        "n_all",
        "agreement_all",
        "n_same_reversal",
        "n_reversal",
        "agreement_reversal",
    ]
    
    print(
        agreement[
            display_columns
        ].to_string(
            index=False
        )
    )
```
