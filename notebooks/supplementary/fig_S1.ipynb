## 8. Retained source analysis and figure generation
The established native calibration/rendering/classical-detector cells below preserve original metrics and plot geometry.


```python
from pathlib import Path
import json, hashlib, sys
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import FancyArrowPatch
sys.path.insert(0,str(ROOT/'src'))
from flyvis_midd.rotation import image_rotation_flow_metrics
DATA=ROOT/'data/precomputed/supplementary/figS1'; OUT=ROOT/'figures/supplementary/figS1'; OUT.mkdir(parents=True,exist_ok=True)
meta=json.loads((DATA/'metadata.json').read_text())
flow=np.load(DATA/'propeller_flow_50models.npy',mmap_mode='r')
with np.load(DATA/'propeller_sequences.npz',allow_pickle=False) as d:
    stimulus_names=d['names'].astype(str); images=d['frames'].astype(np.float32)
with np.load(DATA/'propeller_boxeye_rendered.npz',allow_pickle=False) as d:
    box_names=d['names'].astype(str); luminance=d['luminance'][:,:,0].astype(np.float32); centers_yx=d['receptor_centers_yx'].astype(np.float32)
model_ids=meta['model_ids']; conditions=meta['conditions']; cond_index={name:i for i,name in enumerate(conditions)}
```

## Recompute source metrics with the repository API

```python
source_contract=json.loads((DATA/'source_flow_contract.json').read_text())
source_analysis_contract=json.loads((DATA/'source_analysis_contract.json').read_text())
input_manifest=json.loads((DATA/'source_input_manifest.json').read_text())
source_metrics=pd.read_csv(DATA/'source_flow_metrics.csv',dtype={'model_id':str})
source_control=pd.read_csv(DATA/'source_control_summary.csv',dtype={'model_id':str})
source_pb=pd.read_csv(DATA/'source_P_B_summary.csv',dtype={'model_id':str})
source_qa=pd.read_csv(DATA/'source_analysis_qa.csv')
qa=[]
def check(name,passed,detail=''):
    qa.append({'check':name,'pass':bool(passed),'detail':str(detail)})
def sha256(path):
    h=hashlib.sha256()
    with open(path,'rb') as f:
        for block in iter(lambda:f.read(1024*1024),b''): h.update(block)
    return h.hexdigest()
check('official_50_model_ids_000_049',model_ids==[f'{i:03d}' for i in range(50)] and len(set(model_ids))==50,model_ids[0]+'–'+model_ids[-1])
check('three_expected_conditions',conditions==['Static','CW_mid','CCW_mid'] and stimulus_names.tolist()==conditions and box_names.tolist()==conditions,conditions)
check('flow_expected_shape_and_finite',flow.shape==(50,3,50,2,721) and np.isfinite(flow).all(),flow.shape)
check('stimulus_input_shape',images.shape==(3,50,128,128) and luminance.shape==(3,50,721) and centers_yx.shape==(721,2),(images.shape,luminance.shape,centers_yx.shape))
check('source_artifact_sha256',sha256(DATA/'propeller_flow_50models.npy')==meta['flow']['source_sha256'],meta['flow']['source_sha256'])
check('official_source_QA_pass',pd.read_csv(DATA/'source_flow_qa.csv')['pass'].astype(bool).all() and source_qa['pass'].astype(bool).all(),'both inference and analysis source QA tables')
check('50Hz_fade_1s_frames_10_49',input_manifest['fps']==50 and source_contract['dt_sec']==.02 and source_contract['fade_in_sec']==1.0 and source_contract['stable_frames_python']==[10,50],'zero-based frames 10..49')
check('source_coordinate_contract_one_v_flip',source_contract['flow_coordinate']=='raw decoder (u,v); negate v only for image display' and 'one decoder-v display flip' in source_analysis_contract['rotation_sign'],'image-coordinate tangential metric applies flip internally once')

rows=[]
for mi,mid in enumerate(model_ids):
    for ci,cond in enumerate(conditions):
        raw=np.moveaxis(np.asarray(flow[mi,ci,10:50]),1,2) # (40 frames,721 receptors,2 uv); source frames 10–49
        metric=image_rotation_flow_metrics(centers_yx,raw,inner_radius_fraction=.14,outer_radius_fraction=.98,flip_decoder_y=True)
        rows.append({'model_id':mid,'condition':cond,**{k:v for k,v in metric.items() if k!='annulus_mask'}})
metrics=pd.DataFrame(rows).sort_values(['model_id','condition']).reset_index(drop=True)
check('metrics_unique_50x3',len(metrics)==150 and not metrics.duplicated(['model_id','condition']).any(),metrics.shape)
check('metrics_finite_no_nan_inf',np.isfinite(metrics.select_dtypes('number').to_numpy()).all(),metrics.select_dtypes('number').isna().sum().to_dict())
# Exact reproduction QA against the saved source-analysis per-model table.
joined=metrics.merge(source_metrics,on=['model_id','condition'],suffixes=('_recomputed','_source'),validate='one_to_one')
metric_cols=['mean_flow_magnitude','mean_signed_tangent','signed_tangent_ratio','mean_signed_radial']
metric_diffs={col:float(np.max(np.abs(joined[f'{col}_recomputed']-joined[f'{col}_source']))) for col in metric_cols}
check('local_API_reproduces_source_metrics',max(metric_diffs.values())<2e-5,metric_diffs)
```

## Direction agreement and prespecified detection criterion

```python
# Match the source classify_rotation_control predicate exactly: positive CW, negative CCW;
# then strict ratio thresholds for rotation_flow_detected.
control_rows=[]
for mid,g in metrics.groupby('model_id',sort=True):
    by=g.set_index('condition')
    moving=by.loc[['CW_mid','CCW_mid']]
    aligned=bool(by.loc['CW_mid','signed_tangent_ratio']>0 and by.loc['CCW_mid','signed_tangent_ratio']<0)
    min_abs_ratio=float(np.abs(moving.signed_tangent_ratio).min())
    min_moving_static=float(moving.mean_flow_magnitude.min()/max(float(by.loc['Static','mean_flow_magnitude']),1e-12))
    detected=bool(aligned and min_abs_ratio>.1 and min_moving_static>3.0)
    control_rows.append({'model_id':mid,'CW_signed_ratio':float(by.loc['CW_mid','signed_tangent_ratio']),'CCW_signed_ratio':float(by.loc['CCW_mid','signed_tangent_ratio']),'direction_aligned':aligned,'minimum_absolute_tangent_ratio':min_abs_ratio,'minimum_moving_to_static_magnitude_ratio':min_moving_static,'rotation_flow_detected':detected})
control=pd.DataFrame(control_rows)
source_check=source_control[['model_id','all_labels_aligned','minimum_absolute_tangent_ratio','minimum_moving_to_static_magnitude_ratio','rotation_flow_detected']].copy()
source_check=source_check.rename(columns={'all_labels_aligned':'direction_aligned'})
control_join=control.merge(source_check,on='model_id',suffixes=('_new','_source'),validate='one_to_one')
check('direction_agreement_50_of_50',int(control.direction_aligned.sum())==50, f"{int(control.direction_aligned.sum())}/50")
check('rotation_detection_49_of_50',int(control.rotation_flow_detected.sum())==49,f"{int(control.rotation_flow_detected.sum())}/50")
check('source_classification_reproduced',np.array_equal(control_join.direction_aligned_new,control_join.direction_aligned_source) and np.array_equal(control_join.rotation_flow_detected_new,control_join.rotation_flow_detected_source), 'source criterion thresholds: ratio>0.1 and moving/static magnitude>3.0')
check('CW_positive_and_CCW_negative_all_models',(control.CW_signed_ratio>0).all() and (control.CCW_signed_ratio<0).all(),'direction sign is tested independently of detection')
failed=control.loc[~control.rotation_flow_detected,'model_id'].tolist()
check('non_detected_model_matches_source',['049']==failed,failed)

wide=metrics.pivot(index='model_id',columns='condition',values='mean_signed_tangent').loc[model_ids]
P=(wide.CW_mid-wide.CCW_mid)/2
representative=model_ids[int(np.argmin(np.abs(P.to_numpy()-np.median(P.to_numpy()))))]
source_representative=str(source_pb.loc[(source_pb.P_direction_signal-source_pb.P_direction_signal.median()).abs().idxmin(),'model_id']).zfill(3)
check('representative_is_median_P_model',representative=='005' and representative==source_representative,representative)
```
