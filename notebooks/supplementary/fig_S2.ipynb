## Load bundled source data and verify provenance

Only small saved artifacts are loaded. No FlyVis inference, training, or source-repository access is needed.

```python
import json, hashlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
plt.rcParams.update({'font.size':8,'axes.spines.top':False,'axes.spines.right':False,'figure.dpi':140})
DATA=ROOT/'data/precomputed/supplementary/figS2'
OUT=ROOT/'figures/supplementary/figS2'
OUT.mkdir(parents=True,exist_ok=True)
meta=json.loads((DATA/'metadata.json').read_text())
qa=[]
def check(name,passed,detail=''):
    qa.append({'check':name,'pass':bool(passed),'detail':str(detail)})
def read(name):
    return pd.read_csv(DATA/name,dtype={'model_id':str})
def finite(table):
    return np.isfinite(table.select_dtypes('number').to_numpy()).all()
for record in meta['sources']:
    check('SHA256 '+record['filename'],hashlib.sha256((DATA/record['filename']).read_bytes()).hexdigest()==record.get('copied_sha256', record['source_sha256']))
print(meta['analysis_contract'])
metrics=read('scalar_02_model_level_statistics.csv')
pairs=read('scalar_03_canonical_pair_predictions.csv')
global_metrics=read('scalar_05_global_model_metrics.csv')
field=read('field_01_vector_field_metrics.csv')
representation=read('field_02_representation_level_summary.csv')
renderings=['boxeye','highres_232']
ids=[f'flow/0000/{i:03d}' for i in range(10)]
for table in [metrics,pairs,global_metrics,field,representation]:
    check('model/rendering coverage '+str(table.shape),set(table.model_id)==set(ids) and set(table.rendering)==set(renderings))
    check('finite '+str(table.shape),finite(table))
check('scalar source QA',read('scalar_06_final_QA.csv')['pass'].all())
check('field source QA',read('field_04_vector_field_QA.csv')['pass'].all())
check('exploratory field status',field.analysis_status.eq('unregistered_exploratory').all())
check('canonical pair coverage',len(pairs)==420 and not pairs.duplicated(['model_id','rendering','c1','c2']).any() and (pairs.c1<pairs.c2).all() and pairs.groupby(['model_id','rendering']).size().eq(21).all())
check('full-field pair coverage',len(field)==420 and not field.duplicated(['model_id','rendering','c1','c2']).any() and field.groupby(['model_id','rendering']).size().eq(21).all())
```

## Reproduce source summaries and QA

The source model and rendering contracts remain separate. Calculations below verify saved results; no new inferential tests are introduced.

```python
rows=[]
for (mid,rendering),q in pairs.groupby(['model_id','rendering']):
    x=q.a4_predicted.to_numpy();y=q.a4_observed.to_numpy()
    slope,intercept=np.linalg.lstsq(np.c_[x,np.ones(len(x))],y,rcond=None)[0]
    r2=1-np.sum((y-slope*x-intercept)**2)/np.sum((y-y.mean())**2)
    rmse=np.sqrt(np.mean((y-x)**2))/q.a3_dynamic_range.iloc[0]
    gate=np.abs(x)/q.a3_dynamic_range.to_numpy()>=.05
    check('sign gate '+mid+' '+rendering,np.array_equal(gate,q.sign_evaluable.to_numpy()))
    check('sign flags '+mid+' '+rendering,np.array_equal(gate&(np.sign(x)==np.sign(y)),q.sign_correct.to_numpy()))
    rows.append(dict(model_id=mid,rendering=rendering,slope=slope,intercept=intercept,r2=r2,normalized_rmse=rmse,sign_accuracy=np.mean(np.sign(x[gate])==np.sign(y[gate]))))
reproduced=pd.DataFrame(rows)
for source in [metrics,global_metrics]:
    j=reproduced.merge(source,on=['model_id','rendering'],suffixes=('_new','_source'),validate='one_to_one')
    check('scalar reproduction '+str(source.shape),all(np.allclose(j[c+'_new'],j[c+'_source'],atol=1e-10,rtol=1e-8) for c in ['slope','intercept','r2','normalized_rmse','sign_accuracy']))
agg=field.groupby(['model_id','rendering']).agg(mean_normalized_vector_rmse=('normalized_vector_rmse','mean'),mean_vector_cosine=('cosine_similarity','mean'),mean_vector_correlation=('vector_correlation','mean'),mean_local_tangent_correlation=('local_tangent_correlation','mean')).reset_index()
j=agg.merge(representation,on=['model_id','rendering'],suffixes=('_new','_source'),validate='one_to_one')
check('field summary reproduction',all(np.allclose(j[c+'_new'],j[c+'_source'],atol=1e-10) for c in agg.columns if c not in ['model_id','rendering']))
j=field.merge(pairs,on=['model_id','rendering','c1','c2'],validate='one_to_one')
check('field/global scalar consistency',np.allclose(j.global_tangent_observed,j.a4_observed,atol=1e-8) and np.allclose(j.global_tangent_predicted,j.a4_predicted,atol=1e-8))
summary=metrics.groupby('rendering')[['r2','sign_accuracy','normalized_rmse','slope']].agg(['mean','sem'])
display(summary)
summary.to_csv(OUT/'figS2_scalar_summary.csv')
agg.to_csv(OUT/'figS2_field_model_summary.csv',index=False)
```

## Generate and save the supplementary figure

QA must pass before final figure files are saved.

```python
fig=plt.figure(figsize=(11.2,7.8),constrained_layout=True)
gs=fig.add_gridspec(2,2)
colors=['#4c78a8','#f58518'];labels=['BoxEye','highres_232']
a=fig.add_subplot(gs[0,0]);b=fig.add_subplot(gs[0,1])
for k,(ax,col,title) in enumerate([(a,'r2','A  Fitted scalar R²'),(b,'sign_accuracy','B  Scalar sign agreement')]):
    pivot=metrics.pivot(index='model_id',columns='rendering',values=col)[renderings]
    for row in pivot.to_numpy():ax.plot([0,1],row,color='0.75',lw=.7,zorder=0)
    for i,r in enumerate(renderings):
        v=pivot[r];ax.scatter(np.full(10,i)+np.linspace(-.06,.06,10),v,s=20,color=colors[i]);ax.errorbar(i,v.mean(),yerr=v.sem(),fmt='D',color='black',capsize=3)
        ax.text(i,1.025,f'{v.mean():.3f}',ha='center',fontsize=8)
    ax.set(xticks=[0,1],xticklabels=labels,ylim=(-.03,1.09),title=title,ylabel=col.replace('_',' '))
# Fixed illustrative model, selected independently of goodness of fit.
sub=gs[1,0].subgridspec(1,2)
for i,r in enumerate(renderings):
    ax=fig.add_subplot(sub[0,i]);q=pairs[(pairs.model_id=='flow/0000/000')&(pairs.rendering==r)]
    ax.scatter(q.a4_predicted,q.a4_observed,color=colors[i],s=20)
    lo=min(q.a4_predicted.min(),q.a4_observed.min());hi=max(q.a4_predicted.max(),q.a4_observed.max())
    ax.plot([lo,hi],[lo,hi],ls='--',color='0.5',lw=1)
    m=metrics[(metrics.model_id=='flow/0000/000')&(metrics.rendering==r)].iloc[0]
    x=np.array([lo,hi]);ax.plot(x,m.slope*x+m.intercept,color=colors[i],lw=1)
    ax.set(xlabel='A₄ predicted',ylabel='A₄ observed',title=('C  ' if i==0 else '')+labels[i]+' / model 000')
    ax.ticklabel_format(style='sci',axis='both',scilimits=(-2,2))
    from matplotlib.ticker import MaxNLocator
    ax.xaxis.set_major_locator(MaxNLocator(3))
ax=fig.add_subplot(gs[1,1]);ax2=ax.twinx()
for i,r in enumerate(renderings):
    q=agg[agg.rendering==r];x=i*3
    ax.scatter(np.full(10,x)+np.linspace(-.12,.12,10),q.mean_normalized_vector_rmse,color=colors[i],s=20)
    ax.errorbar(x,q.mean_normalized_vector_rmse.mean(),yerr=q.mean_normalized_vector_rmse.sem(),fmt='D',color='black',capsize=3)
    ax2.scatter(np.full(10,x+1)+np.linspace(-.12,.12,10),q.mean_vector_cosine,color=colors[i],marker='s',s=20)
    ax2.errorbar(x+1,q.mean_vector_cosine.mean(),yerr=q.mean_vector_cosine.sem(),fmt='D',color='black',capsize=3)
ax.set(xticks=[0,1,3,4],xticklabels=['BoxEye\nRMSE','BoxEye\ncosine','highres_232\nRMSE','highres_232\ncosine'],ylabel='Identity vector RMSE / prediction RMS',title='D  Full-field extension (exploratory)')
ax2.set(ylabel='Vector cosine similarity',ylim=(-.35,1.1));ax.axhline(0,color='0.5',lw=.6);ax2.axhline(1,color='0.7',ls=':',lw=.8)
fig.suptitle('S2  Rendering affects scalar composition; scalar fits do not imply field identity',fontsize=11)

qa_table=pd.DataFrame(qa)
qa_table.to_csv(OUT/'figS2_QA.csv',index=False)
display(qa_table)
assert qa_table['pass'].all(), 'QA failed: inspect saved QA before using figure.'
import matplotlib
for text_item in fig.findobj(matplotlib.text.Text):
    if text_item.get_fontsize() < 10:
        text_item.set_fontsize(10)
fig.savefig(OUT/'figS2_main.png',dpi=250,bbox_inches='tight')
fig.savefig(OUT/'figS2_main.pdf',bbox_inches='tight')
plt.show()
plt.close(fig)
print('Saved PNG/PDF and',len(qa_table),'passing QA checks.')
```
