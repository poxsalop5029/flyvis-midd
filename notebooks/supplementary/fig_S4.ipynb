Exploratory comparison. All FlyVis panels use official pretrained model 000 only. ## Load bundled source data and verify provenance

Only small saved artifacts are loaded. No FlyVis inference, training, or source-repository access is needed.

```python
import json, hashlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
plt.rcParams.update({'font.size':11,'axes.spines.top':False,'axes.spines.right':False,'figure.dpi':140})
DATA=ROOT/'data/precomputed/supplementary/figS4'
OUT=ROOT/'figures/supplementary/figS4'
OUT.mkdir(parents=True,exist_ok=True)
meta=json.loads((DATA/'metadata.json').read_text())
qa=[]
def check(name,passed,detail=''):
    qa.append({'check':name,'pass':bool(passed),'detail':str(detail)})
def read(name):
    return pd.read_csv(DATA/name,dtype={'model_id':str})
def finite(table):
    return np.isfinite(table.select_dtypes('number').to_numpy()).all()
for record in meta['sources']:
    check('SHA256 '+record['filename'],hashlib.sha256((DATA/record['filename']).read_bytes()).hexdigest()==record.get('copied_sha256', record['source_sha256']))
print(meta['analysis_contract'])
from scipy.stats import pearsonr,spearmanr
models=read('model_comparison_summary.csv');cal=read('known_rotation_sign_calibration.csv');static=read('calibrated_static_sign_comparison.csv')
land=read('luminance_landscape.csv');decomp=read('mirror_decomposition.csv');rawmetrics=read('landscape_metrics.csv');dirmetrics=read('mirror_directional_metrics.csv');comparison=read('raw_vs_directional_metrics.csv')
check('finite source data',all(finite(x) for x in [models,cal,static,land,decomp,rawmetrics,dirmetrics,comparison]))
check('single-model provenance',meta['analysis_contract']['model_cohort']=='FlyVis official pretrained flow/0000/000 only; classical deterministic detector array')
check('21x21 complete landscape',len(land)==441 and not land.duplicated(['g1','g2']).any() and len(land.g1.unique())==21 and len(land.g2.unique())==21 and np.allclose(sorted(land.g1.unique()),np.linspace(0,1,21)) and np.allclose(sorted(land.g2.unique()),np.linspace(0,1,21)))
check('Bach sign conversion once',cal.bach_to_cw_factor.eq(-1).all() and np.allclose(cal.bach_cw_calibrated,-cal.bach_raw_mean) and np.allclose(land.bach_cw,-land.bach_raw))
check('FlyVis sign conversion once',cal.flyvis_to_cw_factor.eq(1).all() and np.allclose(cal.flyvis_cw_calibrated,cal.flyvis_raw_mean_tangent))
check('CW/CCW correct both models',set(cal.physical_direction)=={'CW','CCW'} and np.array_equal(np.sign(cal.bach_cw_calibrated),cal.cw_positive_target) and np.array_equal(np.sign(cal.flyvis_cw_calibrated),cal.cw_positive_target))
check('static calibrated outputs',np.allclose(static.bach_cw_calibrated,-static.bach_shift_signed) and np.allclose(static.flyvis_cw_calibrated,static.flyvis_mean_signed_tangent) and static.calibrated_sign_agreement.all())
```

## Reproduce source summaries and QA

The source model and rendering contracts remain separate. Calculations below verify saved results; no new inferential tests are introduced.

```python
grid=sorted(land.g1.unique())
B=land.pivot(index='g2',columns='g1',values='bach_cw').loc[grid,grid].to_numpy()
F=land.pivot(index='g2',columns='g1',values='flyvis_cw').loc[grid,grid].to_numpy()
BD=(B-B.T)/2;BS=(B+B.T)/2;FD=(F-F.T)/2;FS=(F+F.T)/2
for col,array in [('bach_directional',BD),('bach_symmetric',BS),('flyvis_directional',FD),('flyvis_symmetric',FS)]:
    saved=decomp.pivot(index='g2',columns='g1',values=col).loc[grid,grid].to_numpy()
    check('mirror identity '+col,np.allclose(saved,array,atol=1e-12))
check('mirror reconstruction',np.allclose(BD+BS,B) and np.allclose(FD+FS,F) and np.allclose(BD,-BD.T) and np.allclose(FD,-FD.T) and np.allclose(np.diag(FD),0))
rows=[]
for name,b,f in [('raw',B,F),('mirror_antisymmetric',BD,FD)]:
    b=b.ravel();f=f.ravel();bn=b/np.max(np.abs(b));fn=f/np.max(np.abs(f));robust=(np.abs(bn)>=.05)&(np.abs(fn)>=.05)
    rows.append(dict(analysis=name,n_conditions=len(b),n_robust=int(robust.sum()),pearson_r=pearsonr(b,f)[0],spearman_rho=spearmanr(b,f)[0],normalized_rmse=np.sqrt(np.mean((bn-fn)**2)),robust_sign_agreement=np.mean(np.sign(b[robust])==np.sign(f[robust]))))
reproduced=pd.DataFrame(rows)
for source in [comparison,rawmetrics.assign(analysis='raw'),dirmetrics]:
    j=reproduced.merge(source,on='analysis',suffixes=('_new','_source'),validate='one_to_one')
    check('metric reproduction '+str(source.shape),all(np.allclose(j[c+'_new'],j[c+'_source'],atol=1e-10) for c in reproduced.columns if c!='analysis'))
rms_a=np.sqrt(np.mean(FD**2));rms_s=np.sqrt(np.mean(FS**2));ratio=rms_s/rms_a
check('RMS reproduction',np.isclose(rms_a,dirmetrics.flyvis_directional_rms.iloc[0]) and np.isclose(rms_s,dirmetrics.flyvis_symmetric_rms.iloc[0]) and np.isclose(ratio,dirmetrics.symmetric_to_directional_rms_ratio.iloc[0]))
check('linear cancellation and nonlinear nonzero',models[models.model.eq('reichardt_array_linear')].mean_abs.lt(1e-12).all() and models[models.model.eq('bach_shift')].mean_abs.gt(1e-6).all())
reproduced.to_csv(OUT/'figS4_reproduced_metrics.csv',index=False);display(reproduced)
print('FlyVis symmetric/directional RMS:',ratio)
```

## Generate and save the supplementary figure

QA must pass before final figure files are saved.

```python
fig=plt.figure(figsize=(11.2,9),constrained_layout=True);gs=fig.add_gridspec(3,2,height_ratios=[1,1.25,.75])
ax=fig.add_subplot(gs[0,0])
for i,stimulus in enumerate(['3midd','4midd']):
    q=models[models.stimulus.eq(stimulus)].set_index('model');values=[q.loc['reichardt_array_linear','mean_abs'],q.loc['bach_shift','mean_abs']];values=np.array(values)/q.loc['bach_shift','mean_abs']
    ax.bar(np.array([0,1])+(i-.5)*.28,values,width=.28,label=stimulus.upper(),color=['#4c78a8','#f58518'][i])
ax.set(xticks=[0,1],xticklabels=['Linear array','Nonlinear array'],ylabel='Absolute output / nonlinear output\n(within stimulus)',ylim=(0,1.25),title='A  Pre-summation nonlinearity prevents cancellation');ax.legend(frameon=False)
ax.text(.03,.66,'Linear absolute outputs < 3×10⁻¹⁷\nShift-average model; not a time course',transform=ax.transAxes,ha='left',fontsize=11)
ax=fig.add_subplot(gs[0,1]);x=np.arange(2)
for i,(col,label,color) in enumerate([('bach_cw_calibrated','Classical nonlinear','#4c78a8'),('flyvis_cw_calibrated','FlyVis model 000','#f58518')]):
    q=cal.set_index('physical_direction').loc[['CW','CCW'],col];ax.bar(x+(i-.5)*.3,q/q.abs().max(),width=.3,color=color,label=label)
ax.axhline(0,color='black',lw=.6);ax.set(xticks=x,xticklabels=['Physical CW','Physical CCW'],ylabel='Within-model normalized output (CW +)',title='B  Known-rotation direction calibration');ax.legend(frameon=False,fontsize=11)
# C: canonical static comparison and amplitude ranking, separately normalized.
ax=fig.add_subplot(gs[1,0])
for i,(col,label,color) in enumerate([('bach_cw_calibrated','Classical nonlinear','#4c78a8'),('flyvis_cw_calibrated','FlyVis model 000','#f58518')]):
    q=static.set_index('stimulus').loc[['3midd','4midd'],col];ax.bar(x+(i-.5)*.3,q/q.abs().max(),width=.3,color=color,label=label)
ax.axhline(0,color='black',lw=.6);ax.set(xticks=x,xticklabels=['3MIDD C=48','4MIDD 85/170'],ylabel='Within-model normalized output (CW +)',title='C  Canonical static outputs / single FlyVis model');ax.legend(frameon=False,fontsize=11)
# D: separate directional landscapes with shared normalized color scale.
sub=gs[1,1].subgridspec(1,2);axes=[fig.add_subplot(sub[0,i]) for i in range(2)]
for ax,z,label in zip(axes,[BD,FD],['Classical','FlyVis model 000']):
    im=ax.imshow(z/np.max(np.abs(z)),origin='lower',extent=[-.025,1.025,-.025,1.025],vmin=-1,vmax=1,cmap='coolwarm',aspect='equal')
    ax.set(xlabel='g₁',ylabel='g₂',title=('D  Antisymmetric\n' if ax is axes[0] else '\n')+label)
fig.colorbar(im,ax=axes,shrink=.75,label='Normalized directional output')
ax=fig.add_subplot(gs[2,0]);names=['Pearson','Spearman','Sign agreement']
for i,name in enumerate(['raw','mirror_antisymmetric']):
    row=reproduced.set_index('analysis').loc[name];v=[row.pearson_r,row.spearman_rho,row.robust_sign_agreement];ax.bar(np.arange(3)+(i-.5)*.3,v,width=.3,label=['Raw','Antisymmetric'][i],color=['#9d9da0','#54a24b'][i])
ax.set(xticks=np.arange(3),xticklabels=names,ylim=(0,1),ylabel='Descriptive landscape metric',title='E  Directional structure comparison');ax.legend(frameon=False,fontsize=11)
ax=fig.add_subplot(gs[2,1]);ax.bar([0,1],[1,ratio],color=['#54a24b','#9d9da0']);ax.set(xticks=[0,1],xticklabels=['Antisymmetric','Symmetric'],ylabel='RMS / directional RMS',title=f'F  Shared output remains substantial ({ratio:.2f}×)')
fig.suptitle('S4  Partial directional agreement / exploratory / FlyVis official model 000 only',fontsize=11)

qa_table=pd.DataFrame(qa)
qa_table.to_csv(OUT/'figS4_QA.csv',index=False)
display(qa_table)
assert qa_table['pass'].all(), 'QA failed: inspect saved QA before using figure.'
import matplotlib
for text_item in fig.findobj(matplotlib.text.Text):
    if text_item.get_fontsize() < 10:
        text_item.set_fontsize(10)
fig.savefig(OUT/'figS4_main.png',dpi=250,bbox_inches='tight')
fig.savefig(OUT/'figS4_main.pdf',bbox_inches='tight')
plt.show()
plt.close(fig)
print('Saved PNG/PDF and',len(qa_table),'passing QA checks.')
```
